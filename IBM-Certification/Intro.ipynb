What is Machine Learning ?

Machine Learning is the subfield of computer science that gives "computers the ability to learn without being explicitly programmed."

Examples:
1. Recommendation.
2. Decision Making
3. Telecommunication customer segmentation etc

Major machine Learning Techiniques:
1. Regression/Estimation
   -> Predicting continuous values, eg->house price prediction,CO2 Emition Pred
2. Classification
   -> Predicting the item class/category of a case
3. Clustering
   -> Finding the structure of data; summarization
4. Associations
   -> Associating frequent co-occuring items/events

---------------Techniques-----------
1. Anomaly detection
   -> Discovering abnormal and unusal cases
2. Sequence Mining
   -> Predicting next events; click-strem(Markov Model, HMM)
3. Dimension Reduction
   -> Reducing the size of data(PCA)
4. Recommendation Systems
   -> Recommendation items
